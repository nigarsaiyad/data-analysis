```python
import numpy as np
```

```python
X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

y = np.array([[0], [1], [1], [0]])
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

# Derivative of Sigmoid

```python
def sigmoid_derivative(x):
    return x * (1 - x)
```

# Initialize Parameters

```python
np.random.seed(42)

input_layer_neurons = 2
hidden_layer_neurons = 3
output_neurons = 1

# Weights
W1 = np.random.uniform(size=(input_layer_neurons, hidden_layer_neurons))
b1 = np.random.uniform(size=(1, hidden_layer_neurons))

W2 = np.random.uniform(size=(hidden_layer_neurons, output_neurons))
b2 = np.random.uniform(size=(1, output_neurons))
```

# Forward Propagation

```python
def forward(X):
    global hidden_input, hidden_output, final_input, output

    hidden_input = np.dot(X, W1) + b1
    hidden_output = sigmoid(hidden_input)

    final_input = np.dot(hidden_output, W2) + b2
    output = sigmoid(final_input)

    return output
```

# Backpropagation

```python
def backward(X, y, output, learning_rate):

    global W1, W2, b1, b2

    # Error
    error = y - output

    # Output layer gradient
    d_output = error * sigmoid_derivative(output)

    # Hidden layer error
    error_hidden = d_output.dot(W2.T)
    d_hidden = error_hidden * sigmoid_derivative(hidden_output)

    # Update weights and biases
    W2 += hidden_output.T.dot(d_output) * learning_rate
    b2 += np.sum(d_output, axis=0, keepdims=True) * learning_rate

    W1 += X.T.dot(d_hidden) * learning_rate
    b1 += np.sum(d_hidden, axis=0, keepdims=True) * learning_rate
```

# Training Loop

```python
learning_rate = 0.1
epochs = 10000

for i in range(epochs):

    output = forward(X)
    backward(X, y, output, learning_rate)

    if i % 1000 == 0:
        loss = np.mean(np.square(y - output))
        print(f"Epoch {i}, Loss: {loss}")
```

# Final Output

```python
print("Final Predictions:")
print(forward(X))
```

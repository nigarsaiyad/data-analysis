```python
import pandas as pd
```

```python
df = pd.read_csv("employee.csv")
df.head()
```

# Great Employee by ID 

```python
def greet_employee(emp_id):
    
    employee = df[df['EID'] == emp_id]

    if employee.empty:
        return "Employee not found"

    full_name = employee.iloc[0]['Employee Name']
    gender = employee.iloc[0]['Gender']

    first_name = full_name.split()[0]

    title = "Mr" if gender == "M" else "Ms"

    return f"Hello, {title} {first_name}"
```

```python
print(greet_employee("E123D"))
```

# Average salary of each dept

```python
def average_salary_by_department():

    avg_salary = (
        df.groupby('Department')['Salary']
        .mean()
        .reset_index()
    )

    return avg_salary


print(average_salary_by_department())
```

#  Employee Details by Employee ID

```python
def employee_details(emp_id):

    employee = df[df['EID'] == emp_id]

    if employee.empty:
        return "Employee not found"

    name = employee.iloc[0]['Employee Name']
    doj = employee.iloc[0]['Date of Joining']
    dept = employee.iloc[0]['Department']

    return f"{name} | {doj} | {dept}"
```

```python
print(employee_details("E123D"))
```

#  Sort Employees from Oldest to Newest

```python
def sort_oldest_to_newest():

    temp_df = df.copy()

    temp_df['Date of Joining'] = pd.to_datetime(
        temp_df['Date of Joining'],
        dayfirst=True
    )

    sorted_df = temp_df.sort_values(
        by='Date of Joining',
        ascending=True
    )

    return sorted_df[
        ['EID',
         'Employee Name',
         'Date of Joining',
         'Department']
    ]


print(sort_oldest_to_newest())
```

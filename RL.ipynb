# Load IMDB Dataset from TensorFlow Datasets

```python
train_data, test_data = tfds.load(
    "imdb_reviews",
    split=["train", "test"],
    as_supervised=True
)
```

# Convert Dataset to Python Lists

```python
test_reviews = []
test_labels = []

for text, label in test_data.take(100):
    test_reviews.append(text.numpy().decode("utf-8"))
    test_labels.append(label.numpy())
```

# Load Transformer Models

```python
distilbert = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)
```

# RoBERTa

```python
roberta = pipeline(
    "sentiment-analysis",
    model="cardiffnlp/twitter-roberta-base-sentiment-latest"
)
```

# Evaluate Models

```python
def evaluate_model(model, reviews, labels):
    
    correct = 0
    
    for review, actual_label in zip(reviews, labels):
        
        result = model(
            review,
            truncation=True,
            max_length=512
        )[0]
        
        prediction = result["label"]
        
        if "POSITIVE" in prediction and actual_label == 1:
            correct += 1
            
        elif "NEGATIVE" in prediction and actual_label == 0:
            correct += 1
            
    return correct / len(reviews)
```

# Compare Accuracy

```python
distilbert_acc = evaluate_model(
    distilbert,
    test_reviews,
    test_labels
)

print("DistilBERT Accuracy:", distilbert_acc)
```

```python
# Conclusion 
Best Model
DistilBERT was chosen as the best model for this assignment because: Fast execution on CPU,
Lower memory usage
Good accuracy on sentiment analysis
Suitable for beginners and small systems
```

# Text Generation using GPT-2

# Load GPT-2

```python
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="gpt2"
)
```

# Define Prompt

```python
prompt = "In a distant future, humanity has discovered"
```

# Generate Story

```python
story = generator(
    prompt,
    max_length=120,
    num_return_sequences=1,
    temperature=0.8,
    truncation=True
)

print(story[0]["generated_text"])
```

```python
train_data = dataset["train"].shuffle(seed=42).select(range(1000))
test_data = dataset["test"].shuffle(seed=42).select(range(200))

print(dataset)
```

# DistilBERT

```python
import warnings
```

```python
warnings.filterwarnings("ignore")

from transformers import pipeline

distilbert_model = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english",
    truncation=True,
    max_length=512
)
```

# BERT sentiment model

```python
bert_model = pipeline(
    "sentiment-analysis",
    model="nlptown/bert-base-multilingual-uncased-sentiment"
)
```

# Test on sample

```python
sample_text = test_data[0]["text"]

print("DistilBERT:", distilbert_model(sample_text))
print("BERT:", bert_model(sample_text))
```

# Simple evaluation

```python
def evaluate(model, data, n=100):
    correct = 0

    for i in range(n):
        text = data[i]["text"]
        label = data[i]["label"]

        pred = model(
            text,
            truncation=True,
            max_length=512
        )[0]["label"]

        if "POSITIVE" in pred and label == 1:
            correct += 1
        elif "NEGATIVE" in pred and label == 0:
            correct += 1

    return correct / n
    
print("DistilBERT Accuracy:", evaluate(distilbert_model, test_data))
print("BERT Accuracy:", evaluate(bert_model, test_data))
```
